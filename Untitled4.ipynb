```python
from google.colab import drive
import pandas as pd

# 1. Montar tu Google Drive (te pedirá autorizar el acceso)
drive.mount('/content/drive')

# 2. Cargar el archivo indicando la ruta dentro de tu Drive
# (Cambia 'Ruta_en_tu_Drive/' por la carpeta exacta donde lo guardaste, por ejemplo: '/content/drive/MyDrive/Carpeta/titanic.csv')
ruta_archivo = '/content/drive/MyDrive/titanic.csv'  # O ajusta la ruta si está en una subcarpeta
df = pd.read_csv(ruta_archivo)

# Verificar que se cargó correctamente
df.head()
```

```python
#¿Cuántos pasajeros hay en el dataset y cuántos PassengerId distintos aparecen? ¿Coinciden los dos resultados?
total_registros = len(df)
passenger_ids_unicos = df['PassengerId'].nunique()

print(f"Total de filas/pasajeros en el dataset: {total_registros}")
print(f"Cantidad de PassengerId distintos: {passenger_ids_unicos}")
print(f"¿Coinciden los resultados?: {'Sí' if total_registros == passenger_ids_unicos else 'No'}")
```

```python
#¿Cuántos pasajeros de cada clase aparecen en el archivo? ¿Qué clase tiene más pasajeros registrados?
pasajeros_por_clase = df['Pclass'].value_counts().sort_index()

print("Cantidad de pasajeros por clase (Pclass):")
print(pasajeros_por_clase)

clase_maxima = pasajeros_por_clase.idxmax()
print(f"\nLa clase con más pasajeros registrados es la Clase {clase_maxima} con {pasajeros_por_clase.max()} pasajeros.")
```

```python
# ¿Cuántos pasajeros embarcaron en cada puerto y cuántos tienen el puerto sin registrar?
# Contamos los valores incluyendo los nulos (NaN)
puertos_conteo = df['Embarked'].value_counts(dropna=False)

print("Desglose de pasajeros por puerto de embarque (C = Cherburgo, Q = Queenstown, S = Southampton, NaN = Sin registrar):")
print(puertos_conteo)

# Extraer explícitamente cuántos no tienen registro
sin_registro = df['Embarked'].isna().sum()
print(f"\nPasajeros con puerto sin registrar: {sin_registro}")
```

```python
#¿Cuál es la tarifa promedio, mínima y máxima registrada para cada clase?
tarifas_por_clase = df.groupby('Pclass')['Fare'].agg(
    Promedio='mean',
    Mínima='min',
    Máxima='max'
).round(2)

print("Tarifas (Fare) por clase de boleto:")
print(tarifas_por_clase)
```

```python
# ¿Cuántos pasajeros de cada clase embarcaron en cada puerto?
# Usamos pd.crosstab para generar la tabla de contingencia con totales por fila y columna
crosstab_clase_puerto = pd.crosstab(df['Pclass'], df['Embarked'], margins=True, margins_name='Total')

print("Pasajeros por Clase y Puerto de Embarque:")
print(crosstab_clase_puerto)
```

```python
# ¿Cuántos pasajeros sobrevivieron y cuántos fallecieron en cada clase?
# Creamos una copia temporal para mapear los números a texto descriptivo en el reporte
df_temp = df.copy()
df_temp['Survival_Status'] = df_temp['Survived'].map({0: 'Falleció', 1: 'Sobrevivió'})

crosstab_supervivencia = pd.crosstab(
    df_temp['Pclass'],
    df_temp['Survival_Status'],
    margins=True,
    margins_name='Total'
)[['Falleció', 'Sobrevivió', 'Total']] # Reordenamos columnas

print("Supervivencia desglosada por clase de boleto:")
print(crosstab_supervivencia)
```

```python
# ¿Cuántos pasajeros viajaban con cero, uno o dos o más familiares registrados a bordo?
# Calculamos el total de familiares a bordo por pasajero
df['FamilySize'] = df['SibSp'] + df['Parch']

# Categorizamos según la consigna
def clasificar_familiares(size):
    if size == 0:
        return '0 familiares'
    elif size == 1:
        return '1 familiar'
    else:
        return '2 o más familiares'

df['FamilyCategory'] = df['FamilySize'].apply(clasificar_familiares)

# Contamos ordenando las categorías lógicamente
conteo_familiares = df['FamilyCategory'].value_counts()[['0 familiares', '1 familiar', '2 o más familiares']]

print("Cantidad de pasajeros según número de familiares a bordo:")
print(conteo_familiares)
```

```python
# ¿Cuántos pasajeros tenían menos de 18 años, cuántos tenían 18 o más y cuántos tenían la edad sin registrar? Desglose el resultado por clase.
# Categorizamos la edad manejando los valores nulos (NaN)
def clasificar_edad(edad):
    if pd.isna(edad):
        return 'Sin registrar'
    elif edad < 18:
        return 'Menores de 18'
    else:
        return '18 o más'

df['AgeCategory'] = df['Age'].apply(clasificar_edad)

# Tabla cruzada por clase y categoría de edad
crosstab_edad_clase = pd.crosstab(
    df['Pclass'],
    df['AgeCategory'],
    margins=True,
    margins_name='Total'
)[['Menores de 18', '18 o más', 'Sin registrar', 'Total']]

print("Desglose de edad por clase de pasajero:")
print(crosstab_edad_clase)
```

```python
# ¿Cuántos códigos de boleto distintos hay? ¿Cuántos aparecen asociados a más de un pasajero y cuál aparece en más registros?
total_tickets_distintos = df['Ticket'].nunique()
conteo_tickets = df['Ticket'].value_counts()

# Boletos que aparecen más de una vez (frecuencia > 1)
tickets_multiples = (conteo_tickets > 1).sum()

# Boleto más frecuente
ticket_mas_frecuente = conteo_tickets.idxmax()
max_apariciones = conteo_tickets.max()

print(f"Total de códigos de boleto distintos: {total_tickets_distintos}")
print(f"Códigos de boleto asociados a más de un pasajero: {tickets_multiples}")
print(f"El boleto que aparece en más registros es '{ticket_mas_frecuente}' con {max_apariciones} pasajeros.")
```

```python
#¿En qué clase hay más registros sin cabina? Para cada clase, muestren cuántos registros tienen cabina, cuántos carecen de ella y cuántos hay en total.
# Creamos una columna auxiliar booleana para verificar si tiene cabina registrada
df['Has_Cabin'] = df['Cabin'].notna()

# Agrupamos por Pclass calculando conteos
resumen_cabinas = df.groupby('Pclass').agg(
    Con_Cabina=('Has_Cabin', lambda x: x.sum()),
    Sin_Cabina=('Has_Cabin', lambda x: (~x).sum()),
    Total=('Pclass', 'count')
)

print("Reporte de registros con y sin cabina por clase:")
print(resumen_cabinas)

clase_mas_sin_cabina = resumen_cabinas['Sin_Cabina'].idxmax()
print(f"\nLa clase con más registros sin cabina es la Clase {clase_mas_sin_cabina} ({resumen_cabinas.loc[clase_mas_sin_cabina, 'Sin_Cabina']} registros).")
```
