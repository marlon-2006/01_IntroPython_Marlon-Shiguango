```python
import pandas as pd
import matplotlib.pyplot as plt


from google.colab import drive
drive.mount('/content/drive')


df = pd.read_csv('/content/drive/MyDrive/WA_Fn-UseC_-HR-Employee-Attrition.csv')

total_empleados = len(df)

tasa_rotacion = (df['Attrition'] == 'Yes').mean() * 100

edad_promedio = df['Age'].mean()

salario_promedio = df['MonthlyIncome'].mean()

antiguedad_promedio = df['YearsAtCompany'].mean()

satisfaccion_promedio = df['JobSatisfaction'].mean()



print("="*60)
print("📊 DASHBOARD DE RECURSOS HUMANOS")
print("="*60)

print(f"👥 Total de empleados: {total_empleados}")
print(f"🔄 Tasa de rotación: {tasa_rotacion:.2f}%")
print(f"🎂 Edad promedio: {edad_promedio:.2f} años")
print(f"💰 Salario promedio: ${salario_promedio:.2f}")
print(f"🏢 Antigüedad promedio: {antiguedad_promedio:.2f} años")
print(f"😊 Satisfacción laboral promedio: {satisfaccion_promedio:.2f}/4")

print("="*60)





print("\nROTACIÓN POR DEPARTAMENTO")

rotacion_departamento = (
    pd.crosstab(df['Department'], df['Attrition'],
                normalize='index') * 100
)

print(rotacion_departamento)



print("\nEMPLEADOS POR DEPARTAMENTO")

print(df['Department'].value_counts())




print("\nSALARIO PROMEDIO POR CARGO")

print(
    df.groupby('JobRole')['MonthlyIncome']
      .mean()
      .sort_values(ascending=False)
)


print("\nSATISFACCIÓN POR CARGO")

print(
    df.groupby('JobRole')['JobSatisfaction']
      .mean()
      .sort_values(ascending=False)
)

# GRAFICO 1

plt.figure(figsize=(8,5))

df['Department'].value_counts().plot(kind='bar')

plt.title('Empleados por Departamento')
plt.xlabel('Departamento')
plt.ylabel('Cantidad')
plt.tight_layout()
plt.show()


plt.figure(figsize=(8,5))

rotacion_departamento['Yes'].plot(kind='bar')

plt.title('Rotación por Departamento')
plt.xlabel('Departamento')
plt.ylabel('% Rotación')
plt.tight_layout()
plt.show()



plt.figure(figsize=(10,6))

df.groupby('JobRole')['MonthlyIncome'] \
  .mean() \
  .sort_values() \
  .plot(kind='barh')

plt.title('Salario Promedio por Cargo')
plt.xlabel('Salario Mensual')
plt.tight_layout()
plt.show()



plt.figure(figsize=(10,6))

df.groupby('JobRole')['JobSatisfaction'] \
  .mean() \
  .sort_values() \
  .plot(kind='barh')

plt.title('Satisfacción Laboral por Cargo')
plt.xlabel('Nivel de Satisfacción')
plt.tight_layout()
plt.show()



plt.figure(figsize=(6,4))

rotacion_overtime = pd.crosstab(
    df['OverTime'],
    df['Attrition'],
    normalize='index'
) * 100

rotacion_overtime['Yes'].plot(kind='bar')

plt.title('Rotación según Horas Extra')
plt.ylabel('% Rotación')
plt.tight_layout()
plt.show()
```
